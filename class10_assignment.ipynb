```python
import pandas as pd

# =====================================================
# 1. FRIEND NAMES
# =====================================================
friends_data = {
    "FriendName": ["Anusha", "prasanna", "meghana", "bhavana", "spandhana", "supraja", "jhansi", "sandhya", "naveena", "roopa"],
    "Quality": ["Intelligent", "Funny", "Intelligent", "Funny", "Intelligent", "Funny", "Intelligent", "Funny", "Intelligent", "Funny"]
}

df_friends = pd.DataFrame(friends_data)
df_friends.to_excel("friend_names.xlsx", index=False)

df_friends = pd.read_excel("friend_names.xlsx")

print("\nIntelligent Friends:")
print(df_friends[df_friends["Quality"] == "Intelligent"])

print("\nFunny Friends:")
print(df_friends[df_friends["Quality"] == "Funny"])


# =====================================================
# 2. FAMILY MEMBERS
# =====================================================
family_data = {
    "FamilyMemberName": ["jayapaul", "rangasarojini", "sunilraju", "salomi", "Raju", "Pooja", "jayaraju", "vijayadurga", "rajesh", "jyothi"],
    "Relation": ["Father", "Mother", "Brother", "Sister", "Brother", "Sister", "Father", "Mother", "Brother", "Sister"]
}

df_family = pd.DataFrame(family_data)
df_family.to_excel("family_members.xlsx", index=False)

df_family = pd.read_excel("family_members.xlsx")

print("\nBrothers & Sisters:")
print(df_family[df_family["Relation"].isin(["Brother", "Sister"])])

print("\nParents:")
print(df_family[df_family["Relation"].isin(["Father", "Mother"])])

print("\nNot Parents:")
print(df_family[~df_family["Relation"].isin(["Father", "Mother"])])

print("\nNot Brothers:")
print(df_family[df_family["Relation"] != "Brother"])

print("\nBrothers OR Parents:")
print(df_family[df_family["Relation"].isin(["Brother", "Father", "Mother"])])

print("\nBrothers AND Parents (No result expected):")
print(df_family[
    (df_family["Relation"] == "Brother") &
    (df_family["Relation"].isin(["Father", "Mother"]))
])


# =====================================================
# 3. VEG FOOD ITEMS
# =====================================================
veg_data = {
    "VegFoodName": ["Tomato Soup", "Veg Curry", "Paneer Fry", "Mushroom Soup", "Aloo Curry", "Gobi Fry", "Dal Soup", "Mix Curry", "Brinjal Fry", "Corn Soup"],
    "Taste": ["Soup", "Curry", "Fry", "Soup", "Curry", "Fry", "Soup", "Curry", "Fry", "Soup"]
}

df_veg = pd.DataFrame(veg_data)
df_veg.to_excel("Vegfood_items.xlsx", index=False)

df_veg = pd.read_excel("Vegfood_items.xlsx")

print("\nSoup Items:")
print(df_veg[df_veg["Taste"] == "Soup"])

print("\nCurry Items:")
print(df_veg[df_veg["Taste"] == "Curry"])

print("\nFry Items:")
print(df_veg[df_veg["Taste"] == "Fry"])

print("\nSoup AND Fry:")
print(df_veg[df_veg["Taste"].isin(["Soup", "Fry"])])

print("\nNOT Soup AND Fry:")
print(df_veg[~df_veg["Taste"].isin(["Soup", "Fry"])])

print("\nEither Soup OR Fry:")
print(df_veg[df_veg["Taste"].isin(["Soup", "Fry"])])


# =====================================================
# 4. NON-VEG FOOD ITEMS
# =====================================================
nonveg_data = {
    "NonVegFoodName": ["Chicken Fry", "Mutton Curry", "Fish Fry", "Chicken Curry", "Mutton Fry", "Egg Fry", "Chicken Roast", "Fish Curry", "Mutton Roast", "Chicken Fry 2"],
    "Taste": ["Fry", "Curry", "Fry", "Curry", "Fry", "Fry", "Fry", "Curry", "Fry", "Fry"]
}

df_nonveg = pd.DataFrame(nonveg_data)
df_nonveg.to_excel("NonVegfood_items.xlsx", index=False)

df_nonveg = pd.read_excel("NonVegfood_items.xlsx")

print("\nChicken Items:")
print(df_nonveg[df_nonveg["NonVegFoodName"].str.contains("Chicken")])

print("\nMutton Items:")
print(df_nonveg[df_nonveg["NonVegFoodName"].str.contains("Mutton")])

print("\nFry Items:")
print(df_nonveg[df_nonveg["Taste"] == "Fry"])

print("\nFry AND Chicken:")
print(df_nonveg[
    (df_nonveg["Taste"] == "Fry") &
    (df_nonveg["NonVegFoodName"].str.contains("Chicken"))
])

print("\nNOT Fry AND Chicken:")
print(df_nonveg[
    ~((df_nonveg["Taste"] == "Fry") &
      (df_nonveg["NonVegFoodName"].str.contains("Chicken")))
])

print("\nEither Fry OR Mutton:")
print(df_nonveg[
    (df_nonveg["Taste"] == "Fry") |
    (df_nonveg["NonVegFoodName"].str.contains("Mutton"))
])


# =====================================================
# 5. MONTH NAMES
# =====================================================
month_data = {
    "MonthName": ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Season": ["Winter", "Winter", "Summer", "Summer", "Summer", "Rainy", "Rainy", "Rainy", "Rainy", "Winter", "Winter", "Winter"]
}

df_month = pd.DataFrame(month_data)
df_month.to_excel("month_names.xlsx", index=False)

df_month = pd.read_excel("month_names.xlsx")

print("\nWinter Months:")
print(df_month[df_month["Season"] == "Winter"])

print("\nSummer Months:")
print(df_month[df_month["Season"] == "Summer"])

print("\nRainy Months:")
print(df_month[df_month["Season"] == "Rainy"])

print("\nWinter AND Summer:")
print(df_month[df_month["Season"].isin(["Winter", "Summer"])])

print("\nNOT Summer AND Rainy:")
print(df_month[~df_month["Season"].isin(["Summer", "Rainy"])])

print("\nEither Winter OR Rainy:")
print(df_month[df_month["Season"].isin(["Winter", "Rainy"])])


# =====================================================
# 6. FLOWER NAMES
# =====================================================
flower_data = {
    "FlowerColour_English": ["RedRose", "WhiteLily", "PinkLotus", "RedTulip", "WhiteJasmine", "PinkRose", "YellowSunflower", "BlueOrchid", "RedDaisy", "PinkLily", "WhiteRose", "PinkTulip"],
    "FlowerColour_Hindi": ["LalGulab", "SafedKumud", "GulabiKamal", "LalTulip", "SafedChameli", "GulabiGulab", "PeelaSurajmukhi", "NeelaOrchid", "LalGulbahar", "GulabiKumud", "SafedGulab", "GulabiTulip"],
    "FlowerColour_Telugu": ["ErraGulabi", "TellaKaluva", "GulabiTamara", "ErraTulip", "TellaMalle", "GulabiGulabi", "PasupuPoddu", "NeelaOrchid", "ErraBanthi", "GulabiKaluva", "TellaGulabi", "GulabiTulip"]
}

df_flower = pd.DataFrame(flower_data)
df_flower.to_excel("flower_names.xlsx", index=False)

df_flower = pd.read_excel("flower_names.xlsx")

print("\nRed Flowers:")
print(df_flower[df_flower["FlowerColour_English"].str.contains("Red")])

print("\nWhite Flowers:")
print(df_flower[df_flower["FlowerColour_English"].str.contains("White")])

print("\nPink Flowers:")
print(df_flower[df_flower["FlowerColour_English"].str.contains("Pink")])

print("\nRed AND Pink:")
print(df_flower[df_flower["FlowerColour_English"].str.contains("Red|Pink")])

print("\nNOT Red AND White:")
print(df_flower[~df_flower["FlowerColour_English"].str.contains("Red|White")])

print("\nEither Red OR Pink:")
print(df_flower[df_flower["FlowerColour_English"].str.contains("Red|Pink")])
```
